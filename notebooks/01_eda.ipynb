```python
import pandas as pd
df = pd.read_csv("data/RAW/pump_telegram.csv")
print(df.shape)
print(df.columns.tolist())
print(df.head(3))
```

```python
# La Morgia stores base symbol only (e.g. 'BRD', not 'BRDBTC').
# In 2018-2021 Binance, virtually all pump targets were /BTC pairs.
# The /BTC filter is therefore equivalent to Binance-only for this dataset.
filt = df[df['exchange'].str.lower() == 'binance'].copy()
print(f"Total events: {len(df)}, Binance (/BTC proxy) events: {len(filt)}")
```

```python
# hour column is already 'HH:MM'; append ':00' to form a full timestamp
filt['pump_ts'] = pd.to_datetime(
    filt['date'] + ' ' + filt['hour'].astype(str) + ':00',
    utc=True
)
print(filt['pump_ts'].min(), filt['pump_ts'].max())
```

```python
import matplotlib.pyplot as plt
filt['year'] = filt['pump_ts'].dt.year
filt['year'].value_counts().sort_index().plot(kind='bar')
plt.title('Pumps per Year (Binance)')
plt.xlabel('Year')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
filt['symbol'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Most Pumped Symbols')
plt.xlabel('Symbol')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
filt['pump_ts'].dt.hour.value_counts().sort_index().plot(kind='bar')
plt.title('Pump Hour Distribution (UTC)')
plt.xlabel('Hour (UTC)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
